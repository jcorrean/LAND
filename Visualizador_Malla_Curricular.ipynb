```python
# ==========================================
# GRAFO ASIGNATURAS -> COMPETENCIAS
# CON FUNCIONES Y TOOLTIPS
# ==========================================

!pip install pyvis

import json
import pandas as pd
from pyvis.network import Network

# ==========================================
# 1. LEER ARCHIVO DE ARISTAS
# ==========================================

df = pd.read_csv("/content/LAND/Lista de aristas.csv", sep=',', encoding='utf8')

# ==========================================
# 2. DEFINICIÓN DE FUNCIONES
# ==========================================

funciones = {

    "1": "Modelos de negocio y emprendimiento digital",
    "2": "Analítica de datos y humanización del dato",
    "3": "Toma de decisiones y resolución de problemas",
    "4": "Liderazgo humanista y transformación digital",
    "5": "Transformación y sostenibilidad organizacional"
}

# ==========================================
# 3. DESCRIPCIONES DE COMPETENCIAS
# ==========================================

descripciones = {

"Competencia 1.1":
"Propone modelos de negocio y emprendimiento en sectores digitales para capitalizar oportunidades estratégicas.",

"Competencia 1.2":
"Propone nuevos modelos de negocio y apoya a la alta dirección en la toma de decisiones.",

"Competencia 1.3":
"Implementa estrategias de emprendimiento digital para generar valor económico y social.",

"Competencia 2.1":
"Procesa grandes volúmenes de datos mediante software estadístico y lenguajes de programación.",

"Competencia 2.2":
"Identifica patrones, correlaciones y tendencias para la toma de decisiones estratégicas.",

"Competencia 2.3":
"Diseña estrategias de intervención sustentadas en datos para optimizar procesos organizacionales.",

"Competencia 3.1":
"Desarrolla KPI e informes mediante integración de datos comerciales, financieros y operativos.",

"Competencia 3.2":
"Resuelve problemas complejos utilizando pensamiento sistémico y evidencia.",

"Competencia 3.3":
"Aplica criterios éticos en la gestión y análisis de datos.",

"Competencia 4.1":
"Identifica sesgos algorítmicos y riesgos de exclusión o discriminación.",

"Competencia 4.2":
"Evalúa críticamente la calidad de los datos para proponer soluciones innovadoras.",

"Competencia 4.3":
"Lidera proyectos tecnológicos considerando impactos ambientales, sociales y de negocio.",

"Competencia 5.1":
"Diseña estrategias de transformación digital con atención a impactos ambientales y sociales.",

"Competencia 5.2":
"Asesora a la alta dirección mediante comunicación efectiva basada en evidencia.",

"Competencia 5.3":
"Aplica modelos de gestión de riesgos y continuidad de negocio."
}

# ==========================================
# 4. COLORES POR FUNCIÓN
# ==========================================

colores_funcion = {

    "1": "#1f77b4",   # azul
    "2": "#2ca02c",   # verde
    "3": "#ff7f0e",   # naranja
    "4": "#9467bd",   # morado
    "5": "#d62728"    # rojo
}

# ==========================================
# 5. INICIALIZAR RED
# ==========================================

net = Network(
    height="900px",
    width="100%",
    bgcolor="#222222",
    font_color="white",
    notebook=True
)

# ==========================================
# 6. NODOS ASIGNATURAS
# ==========================================

asignaturas = sorted(df["Asignatura"].unique())

for asignatura in asignaturas:

    grado = len(
        df[df["Asignatura"] == asignatura]
    )

    net.add_node(
        asignatura,
        label=asignatura,
        color="#4A90E2",
        tipo="asignatura",
        shape="dot",
        size=15 + grado,
        title=f"""
        <b>Asignatura</b><br>
        {asignatura}<br><br>
        Competencias asociadas: {grado}
        """
    )

# ==========================================
# 7. NODOS COMPETENCIAS
# ==========================================

competencias = sorted(
    df["Competencia_ID"].unique()
)

for comp in competencias:

    grado = len(
        df[df["Competencia_ID"] == comp]
    )

    funcion = comp.split()[1][0]

    tooltip = f"""
    <b>{comp}</b><br><br>

    <b>Función {funcion}</b><br>
    {funciones[funcion]}<br><br>

    <b>Descripción</b><br>
    {descripciones[comp]}<br><br>

    Asignaturas conectadas: {grado}
    """

    net.add_node(
        comp,
        label=comp.replace("Competencia ", ""),
        shape="diamond",
        color=colores_funcion[funcion],
        size=18 + grado * 2,
        funcion=funcion,
        tipo="competencia",
        title=tooltip
)
    nodes_data = []

for n in net.nodes:

    nodes_data.append({

        "id": n["id"],

        "tipo": n.get("tipo", ""),

        "funcion": n.get("funcion", "")

    })
# ==========================================
# 8. ARISTAS
# ==========================================

for _, row in df.iterrows():

    net.add_edge(
        row["Asignatura"],
        row["Competencia_ID"],
        color="#999999",
        width=1.5
    )

# ==========================================
# 9. OPCIONES VISUALES
# ==========================================

net.set_options("""
{
  "interaction": {

    "hover": true,
    "tooltipDelay": 200,
    "navigationButtons": true,
    "dragNodes": true,
    "multiselect": true,
    "zoomView": true

  },

  "physics": {

    "barnesHut": {

      "gravitationalConstant": -30000,
      "centralGravity": 0.15,
      "springLength": 180,
      "springConstant": 0.04

    },

    "stabilization": {

      "iterations": 200

    }
  },

  "nodes": {

     "font": {
        "size": 14,
        "color": "white"
     }

  }
}
""")

# ==========================================
# 10. GUARDAR HTML
# ==========================================

net.save_graph("grafo_competencias.html")

# ==========================================
# 11. PANEL IZQUIERDO
# ==========================================

selector_html = """
<div id="panel" style="
position:absolute;
top:20px;
left:20px;
width:340px;
background:rgba(0,0,0,0.9);
padding:20px;
border-radius:10px;
border:1px solid gray;
color:white;
z-index:999;
font-family:Arial;
font-size:12px;
">

<h2>Explorador Curricular</h2>

<label><b>Asignatura</b></label><br>

<select id="asignaturaFilter"
style="width:100%;"
onchange="applyFilters()">

<option value="ALL">Todas</option>
"""

for a in asignaturas:

    selector_html += f"""
    <option value="{a}">
    {a}
    </option>
    """

selector_html += """

</select>

<br><br>

<label><b>Función</b></label><br>

<select id="funcionFilter"
style="width:100%;"
onchange="applyFilters()">

<option value="ALL">Todas</option>
<option value="1">Función 1</option>
<option value="2">Función 2</option>
<option value="3">Función 3</option>
<option value="4">Función 4</option>
<option value="5">Función 5</option>

</select>

<br><br>

<label><b>Competencia</b></label><br>

<select id="competenciaFilter"
style="width:100%;"
onchange="applyFilters()">

<option value="ALL">Todas</option>
"""

for c in competencias:

    selector_html += f"""
    <option value="{c}">
    {c}
    </option>
    """

selector_html += """

</select>

<br><br>

<button onclick="resetGraph()">
Mostrar Red Completa
</button>

</div>
"""

# ==========================================
# 12. LEYENDA DERECHA
# ==========================================

legend_html = """
<div id="legend" style="
position:absolute;
top:20px;
right:20px;
width:360px;
max-height:90%;
overflow-y:auto;
background:rgba(0,0,0,0.9);
padding:20px;
border-radius:10px;
border:1px solid gray;
z-index:999;
color:white;
font-family:Arial;
font-size:12px;
">

<h2>Funciones y Competencias</h2>

<p><span style="color:#4A90E2;">●</span>
Asignaturas
</p>

<hr>

<p><span style="color:#1f77b4;">◆</span>
<b>Función 1</b><br>
Modelos de negocio y emprendimiento digital
</p>

<p><span style="color:#2ca02c;">◆</span>
<b>Función 2</b><br>
Analítica de datos y humanización del dato
</p>

<p><span style="color:#ff7f0e;">◆</span>
<b>Función 3</b><br>
Toma de decisiones y resolución de problemas
</p>

<p><span style="color:#9467bd;">◆</span>
<b>Función 4</b><br>
Liderazgo humanista y transformación digital
</p>

<p><span style="color:#d62728;">◆</span>
<b>Función 5</b><br>
Transformación y sostenibilidad organizacional
</p>

<hr>

<p>
Use los filtros para explorar el currículo.
</p>

</div>
"""

# ==========================================
# 13. JAVASCRIPT
# ==========================================

nodes_json = json.dumps(nodes_data)

js_code = f"""

<script>

var nodesData = {nodes_json};

function resetGraph(){{

    document.getElementById(
        "asignaturaFilter"
    ).value="ALL";

    document.getElementById(
        "funcionFilter"
    ).value="ALL";

    document.getElementById(
        "competenciaFilter"
    ).value="ALL";

    applyFilters();
}}

function applyFilters(){{

    var asignatura =
        document.getElementById(
        "asignaturaFilter"
        ).value;

    var funcion =
        document.getElementById(
        "funcionFilter"
        ).value;

    var competencia =
        document.getElementById(
        "competenciaFilter"
        ).value;

    nodes.forEach(function(n){{
        nodes.update({{
            id:n.id,
            hidden:true
        }});
    }});

    edges.forEach(function(e){{
        edges.update({{
            id:e.id,
            hidden:true
        }});
    }});

    if(
        asignatura=="ALL"
        &&
        funcion=="ALL"
        &&
        competencia=="ALL"
    ){{

        nodes.forEach(function(n){{
            nodes.update({{
                id:n.id,
                hidden:false
            }});
        }});

        edges.forEach(function(e){{
            edges.update({{
                id:e.id,
                hidden:false
            }});
        }});

        network.fit();

        return;
    }}

    let visibleNodes = new Set();

    edges.forEach(function(edge){{

        let mostrar = false;

        if(
            asignatura!="ALL"
            &&
            (
             edge.from==asignatura
             ||
             edge.to==asignatura
            )
        ) mostrar=true;

        if(
            competencia!="ALL"
            &&
            (
             edge.from==competencia
             ||
             edge.to==competencia
            )
        ) mostrar=true;

        if(funcion!="ALL"){{

            nodesData.forEach(function(n){{

                if(
                    n.funcion==funcion
                    &&
                    (
                     edge.from==n.id
                     ||
                     edge.to==n.id
                    )
                ){{
                    mostrar=true;
                }}

            }});
        }}

        if(mostrar){{

            edges.update({{
                id:edge.id,
                hidden:false
            }});

            visibleNodes.add(edge.from);

            visibleNodes.add(edge.to);
        }}

    }});

    nodes.forEach(function(node){{

        nodes.update({{
            id:node.id,
            hidden:!visibleNodes.has(node.id)
        }});

    }});

    network.fit({{
        animation:true
    }});

}}

</script>

"""

# ==========================================
# 14. INYECTAR AL HTML
# ==========================================

with open(
    "grafo_competencias.html",
    "r",
    encoding="utf-8"
) as f:

    html = f.read()

html = html.replace(
    "<body>",
    f"<body>{selector_html}{legend_html}"
)

html = html.replace(
    "</body>",
    js_code + "</body>"
)

with open(
    "grafo_competencias.html",
    "w",
    encoding="utf-8"
) as f:

    f.write(html)

print("Grafo generado correctamente.")

print("Grafo generado correctamente.")
```
